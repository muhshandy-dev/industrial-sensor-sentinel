```python
import json

repo = json.load(open("/kaggle/working/repo/reports/results.json"))
kaggle = json.load(open("/kaggle/working/results_kaggle.json"))
rows = []
for protokol, models in kaggle["protokol"].items():
    for model, modes in models.items():
        for mode, s in modes.items():
            r = repo["protokol"][protokol][model][mode]
            rows.append(
                (
                    protokol,
                    model,
                    mode,
                    round(s["uji"]["f1"], 4),
                    round(r["uji"]["f1"], 4),
                    abs(s["uji"]["f1"] - r["uji"]["f1"]) < 1e-6,
                )
            )
for row in rows:
    print(row)
print("Semua sama dengan hasil di repo:", all(r[-1] for r in rows))
```
